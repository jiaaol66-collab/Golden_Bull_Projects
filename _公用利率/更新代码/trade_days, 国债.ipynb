```python
import exchange_calendars as ecals
import pandas as pd
```

```python
YR = 2026
```

```python
df_bonds = [pd.read_excel(f'../国债/{i}年中债国债收益率曲线标准期限信息.xlsx') for i in range(YR - 7, YR)]
```

```python
df_bond = pd.concat(df_bonds)
```

```python
df_bond.info()
```

```python
df_bond['标准期限说明'].value_counts().index.to_list()
```

```python
with pd.ExcelWriter('../2026/国债分年期数据.xlsx') as writer:
    for term in df_bond['标准期限说明'].unique():
        df_bond[df_bond['标准期限说明'] == term].sort_values(by=['日期']).to_excel(writer, sheet_name = term, index=False)
```

```python
cn_calendar = ecals.get_calendar("XSHG")  # 上海证券交易所 (XSHG)
# 获取 2019-01-01 到 2023-12-31 的交易日
trade_days = cn_calendar.sessions_in_range(pd.Timestamp(f'{YR-7}-01-01'), pd.Timestamp(f'{YR-1}-12-31'))
trade_days = pd.DataFrame({'交易日期':trade_days})
trade_days.info()
trade_days['交易日期'] = trade_days['交易日期'].dt.strftime('%Y-%m-%d')
```
